```python
import sqlite3

conn = sqlite3.connect("inventory.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    product_code TEXT UNIQUE,
    product_name TEXT NOT NULL,
    category TEXT,
    unit TEXT NOT NULL,
    purchase_price REAL,
    selling_price REAL,
    stock REAL DEFAULT 0,
    minimum_stock REAL DEFAULT 0
)
""")

conn.commit()

print("Database created successfully!")
```

```python
cursor.execute("SELECT * FROM products")

products = cursor.fetchall()

for product in products:
    print(product)
```

```python
def add_product():
    code = input("Product Code: ")
    name = input("Product Name: ")
    category = input("Category: ")
    unit = input("Unit (Piece/Kg/Gram/Liter/Box): ")
    purchase = float(input("Purchase Price: "))
    selling = float(input("Selling Price: "))
    stock = float(input("Opening Stock: "))
    minimum = float(input("Minimum Stock: "))

    try:
        cursor.execute("""
        INSERT INTO products
        (product_code, product_name, category, unit,
         purchase_price, selling_price, stock, minimum_stock)
        VALUES (?, ?, ?, ?, ?, ?, ?, ?)
        """, (code, name, category, unit, purchase, selling, stock, minimum))

        conn.commit()
        print("Product added successfully!")

    except sqlite3.IntegrityError:
        print("Error: Product code already exists.")
```

```python
def stock_in():
    code = input("Product Code: ")
    qty = float(input("Quantity received: "))

    cursor.execute(
        "UPDATE products SET stock = stock + ? WHERE product_code = ?",
        (qty, code)
    )

    cursor.execute("""
        INSERT INTO transactions
        (product_code, transaction_type, quantity, date)
        VALUES (?, 'IN', ?, datetime('now'))
    """, (code, qty))

    conn.commit()
    print("Stock added successfully!")
stock_in()
```

```python
def stock_out():
    code = input("Product Code: ")
    qty = float(input("Quantity issued/sold: "))

    cursor.execute(
        "SELECT stock FROM products WHERE product_code = ?",
        (code,)
    )

    result = cursor.fetchone()

    if result is None:
        print("Product not found!")
        return

    if qty > result[0]:
        print("Insufficient stock!")
        return

    cursor.execute(
        "UPDATE products SET stock = stock - ? WHERE product_code = ?",
        (qty, code)
    )

    cursor.execute("""
        INSERT INTO transactions
        (product_code, transaction_type, quantity, date)
        VALUES (?, 'OUT', ?, datetime('now'))
    """, (code, qty))

    conn.commit()
    print("Stock removed successfully!")
stock_out()
```

```python
import pandas as pd
def view_inventory():
    cursor.execute("""
        SELECT product_code, product_name, category, unit,
               purchase_price, selling_price, stock, minimum_stock
        FROM products
    """)

    data = cursor.fetchall()

    columns = [
        "Code", "Product", "Category", "Unit",
        "Purchase Price", "Selling Price",
        "Stock", "Minimum Stock"
    ]

    df = pd.DataFrame(data, columns=columns)
    display(df)
view_inventory()
```

```python
def low_stock():
    cursor.execute("""
        SELECT product_code, product_name, unit, stock, minimum_stock
        FROM products
        WHERE stock <= minimum_stock
    """)

    data = cursor.fetchall()

    df = pd.DataFrame(
        data,
        columns=["Code", "Product", "Unit", "Stock", "Minimum Stock"]
    )

    display(df)
low_stock()
```

```python
import sqlite3
import pandas as pd

# Connect to database and initialize cursor
conn = sqlite3.connect("inventory.db")
cursor = conn.cursor()

print("Database connected successfully!")
```

```python
import pandas as pd

def export_to_excel():
    cursor.execute("""
        SELECT product_code, product_name, category, unit,
               purchase_price, selling_price, stock, minimum_stock
        FROM products
    """)

    data = cursor.fetchall()

    if not data:
        print("No product records found to export.")
        return

    columns = [
        "Code", "Product", "Category", "Unit",
        "Purchase Price", "Selling Price",
        "Stock", "Minimum Stock"
    ]

    df = pd.DataFrame(data, columns=columns)

    # Calculate stock financial values
    df["Total Stock Value (Cost)"] = df["Stock"] * df["Purchase Price"]
    df["Total Potential Revenue"] = df["Stock"] * df["Selling Price"]

    # Export to Excel file
    file_name = "inventory_report.xlsx"
    df.to_excel(file_name, index=False)

    print(f"✓ Inventory exported to '{file_name}' successfully!")

# Run function
export_to_excel()
```

```python
cursor.execute("DROP TABLE IF EXISTS transactions")

cursor.execute("""
CREATE TABLE transactions (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    product_code TEXT,
    transaction_type TEXT,
    quantity REAL,
    unit_price REAL,
    party TEXT,
    remarks TEXT,
    date TEXT
)
""")

conn.commit()

print("Transaction table updated!")
```

```python
def stock_in():
    code = input("Product Code: ")
    qty = float(input("Quantity received: "))
    price = float(input("Purchase Price: "))
    party = input("Supplier: ")
    remarks = input("Remarks: ")

    cursor.execute(
        "SELECT stock FROM products WHERE product_code = ?",
        (code,)
    )

    if cursor.fetchone() is None:
        print("Product not found!")
        return

    cursor.execute(
        "UPDATE products SET stock = stock + ? WHERE product_code = ?",
        (qty, code)
    )

    cursor.execute("""
        INSERT INTO transactions
        (product_code, transaction_type, quantity,
         unit_price, party, remarks, date)
        VALUES (?, 'IN', ?, ?, ?, ?, datetime('now'))
    """, (code, qty, price, party, remarks))

    conn.commit()

    print("Stock received successfully!")
def stock_out():
    code = input("Product Code: ")
    qty = float(input("Quantity issued/sold: "))
    price = float(input("Selling Price: "))
    party = input("Customer: ")
    remarks = input("Remarks: ")

    cursor.execute(
        "SELECT stock FROM products WHERE product_code = ?",
        (code,)
    )

    result = cursor.fetchone()

    if result is None:
        print("Product not found!")
        return

    if qty > result[0]:
        print("Insufficient stock!")
        return

    cursor.execute(
        "UPDATE products SET stock = stock - ? WHERE product_code = ?",
        (qty, code)
    )

    cursor.execute("""
        INSERT INTO transactions
        (product_code, transaction_type, quantity,
         unit_price, party, remarks, date)
        VALUES (?, 'OUT', ?, ?, ?, ?, datetime('now'))
    """, (code, qty, price, party, remarks))

    conn.commit()

    print("Stock issued successfully!")
def transaction_history():
    cursor.execute("""
        SELECT id, product_code, transaction_type,
               quantity, unit_price, party, remarks, date
        FROM transactions
        ORDER BY id DESC
    """)

    data = cursor.fetchall()

    columns = [
        "ID", "Code", "Type", "Quantity",
        "Unit Price", "Party", "Remarks", "Date"
    ]

    df = pd.DataFrame(data, columns=columns)
    display(df)
transaction_history()
```

```python
def dashboard():
    # Total products
    cursor.execute("SELECT COUNT(*) FROM products")
    total_products = cursor.fetchone()[0]

    # Total stock
    cursor.execute("SELECT SUM(stock) FROM products")
    total_stock = cursor.fetchone()[0] or 0

    # Total stock value
    cursor.execute("""
        SELECT SUM(stock * purchase_price)
        FROM products
    """)
    stock_value = cursor.fetchone()[0] or 0

    # Low stock count
    cursor.execute("""
        SELECT COUNT(*)
        FROM products
        WHERE stock <= minimum_stock
    """)
    low_stock_count = cursor.fetchone()[0]

    print("\n===================================")
    print("        INVENTORY DASHBOARD")
    print("===================================")
    print(f"Total Products   : {total_products}")
    print(f"Total Stock      : {total_stock}")
    print(f"Stock Value      : Rs. {stock_value:,.2f}")
    print(f"Low Stock Items  : {low_stock_count}")
    print("===================================")
dashboard()
```

```python
def main_menu():
    while True:
        print("\n==============================")
        print(" INVENTORY MANAGEMENT SYSTEM")
        print("==============================")
        print("1. Dashboard")
        print("2. Add Product")
        print("3. View Inventory")
        print("4. Stock In")
        print("5. Stock Out")
        print("6. Search Product")
        print("7. Low Stock")
        print("8. Transaction History")
        print("9. Export to Excel")
        print("10. Exit")
        print("==============================")

        choice = input("Enter your choice: ")

        if choice == "1":
            dashboard()
        elif choice == "2":
            add_product()
        elif choice == "3":
            view_inventory()
        elif choice == "4":
            stock_in()
        elif choice == "5":
            stock_out()
        elif choice == "6":
            search_product()
        elif choice == "7":
            low_stock()
        elif choice == "8":
            transaction_history()
        elif choice == "9":
            export_to_excel()
        elif choice == "10":
            print("Thank you for using the system!")
            break
        else:
            print("Invalid choice.")
main_menu()
```

```python
def edit_product():
    code = input("Enter Product Code: ")

    cursor.execute(
        "SELECT * FROM products WHERE product_code = ?",
        (code,)
    )

    product = cursor.fetchone()

    if product is None:
        print("Product not found!")
        return

    print("Leave blank to keep the existing value.")

    name = input(f"Product Name [{product[2]}]: ") or product[2]
    category = input(f"Category [{product[3]}]: ") or product[3]
    unit = input(f"Unit [{product[4]}]: ") or product[4]

    purchase = input(f"Purchase Price [{product[5]}]: ")
    selling = input(f"Selling Price [{product[6]}]: ")
    minimum = input(f"Minimum Stock [{product[8]}]: ")

    purchase = float(purchase) if purchase else product[5]
    selling = float(selling) if selling else product[6]
    minimum = float(minimum) if minimum else product[8]

    cursor.execute("""
        UPDATE products
        SET product_name = ?,
            category = ?,
            unit = ?,
            purchase_price = ?,
            selling_price = ?,
            minimum_stock = ?
        WHERE product_code = ?
    """, (
        name, category, unit,
        purchase, selling, minimum, code
    ))

    conn.commit()

    print("Product updated successfully!")
edit_product()
```

```python
def delete_product():
    code = input("Enter Product Code to delete: ")

    cursor.execute(
        "SELECT product_name FROM products WHERE product_code = ?",
        (code,)
    )

    product = cursor.fetchone()

    if product is None:
        print("Product not found!")
        return

    confirm = input(
        f"Delete {product[0]}? (yes/no): "
    ).lower()

    if confirm == "yes":
        cursor.execute(
            "DELETE FROM products WHERE product_code = ?",
            (code,)
        )

        conn.commit()

        print("Product deleted successfully!")
    else:
        print("Delete cancelled.")
print("1. Dashboard")
print("2. Add Product")
print("3. View Inventory")
print("4. Edit Product")
print("5. Delete Product")
print("6. Stock In")
print("7. Stock Out")
print("8. Search Product")
print("9. Low Stock")
print("10. Transaction History")
print("11. Export to Excel")
print("12. Exit")

```

```python
import matplotlib.pyplot as plt

def show_charts():
    global cursor
    cursor.execute("SELECT product_name, stock, (stock * purchase_price) AS stock_value FROM products")
    data = cursor.fetchall()
    
    if not data:
        print("\nNo product data available to plot.")
        return

    df = pd.DataFrame(data, columns=["Product", "Stock Quantity", "Stock Value"])

    # Create a layout with 2 subplots side-by-side
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

    # --- Chart 1: Bar Chart for Stock Quantity ---
    ax1.bar(df["Product"], df["Stock Quantity"], color="skyblue", edgecolor="black")
    ax1.set_title("Stock Quantity per Product", fontsize=14, fontweight="bold")
    ax1.set_xlabel("Product", fontsize=11)
    ax1.set_ylabel("Quantity", fontsize=11)
    ax1.tick_params(axis='x', rotation=30)
    ax1.grid(axis='y', linestyle='--', alpha=0.7)

    # --- Chart 2: Pie Chart for Stock Value Distribution ---
    ax2.pie(
        df["Stock Value"], 
        labels=df["Product"], 
        autopct="%1.1f%%", 
        startangle=140, 
        colors=plt.cm.Paired.colors
    )
    ax2.set_title("Stock Value Share (%)", fontsize=14, fontweight="bold")

    plt.tight_layout()
    plt.show()

def show_transaction_chart():
    global cursor
    cursor.execute("""
        SELECT transaction_type, SUM(quantity) 
        FROM transactions 
        GROUP BY transaction_type
    """)
    data = cursor.fetchall()

    if not data:
        print("\nNo transaction history available to plot.")
        return

    df_trans = pd.DataFrame(data, columns=["Type", "Total Quantity"])

    
    plt.figure(figsize=(6, 6))
    plt.pie(
        df_trans["Total Quantity"], 
        labels=df_trans["Type"], 
        autopct="%1.1f%%", 
        colors=["#2ecc71", "#e74c3c"], 
        startangle=90,
        explode=(0.05, 0) if len(df_trans) > 1 else None
    )
    plt.title("Stock In vs. Stock Out Share", fontsize=14, fontweight="bold")
    plt.show()
show_charts()
```
